```python
#Importando mismo data frame para uso de numpy
import numpy as np
import pandas as pd

dataframe_url = 'https://www.datos.gov.co/resource/sfcd-b3ey.json'
df = pd.read_json(dataframe_url)
df.head()
```

```python
import numpy as np
import pandas as pd

#generar arreglos y obtener el promedio
array_from_df = df['propiedadepm']
array_of_propiedadepm = array_from_df.to_numpy()
suma = np.nansum(array_of_propiedadepm)
print(suma)
```

```python
import numpy as np
import pandas as pd

#generar arreglos y obtener la media
array_from_df = df['propiedadepm']
array_of_propiedadepm = array_from_df.to_numpy()
suma = np.nanmean(array_of_propiedadepm)
print(suma)
```

```python
import numpy as np
import pandas as pd

#generar arreglos y matrices
array_from_df = df['propiedadepm']
array_of_propiedadepm = array_from_df.to_numpy()
suma = np.nanmedian(array_of_propiedadepm)
print(suma)
```

```python
import numpy as np
import pandas as pd

#generar matrices
array_propiedadepm_from_df = df['propiedadepm']
array_compartidos_from_df = df['compartido']
array_of_propiedadepm = array_from_df.to_numpy()
array_of_compartidos = array_compartidos_from_df.to_numpy()
matrix = np.array((array_of_propiedadepm, array_of_compartidos))
print(matrix)
```

```python
#shape
print(matrix.shape)
```

```python
#for anidado para calcular diferencia entre propiedadepm y compratido de matrix
for i in range(len(matrix)):
    for j in range(len(matrix)):
      print('diferencia', np.abs(matrix[i][0] - matrix[j][0]))
```
