```python
import pandas as pd
dataframe_url = 'https://www.datos.gov.co/resource/a7iv-sme8.json'
df = pd.read_json(dataframe_url)
df.head()
```

```python
#Valores asociados a cadenas de texto
conteo_principio_activo = df['principioactivo'].value_counts()
print(conteo_principio_activo) #Array por clave & valor
```

```python
#Valores Asociados a forma farmaceutica
forma_contador = df['formafarmaceutica'].value_counts()
print(forma_contador)
```

```python
print("NORMALIZACION DE VALORES")
df['principioactivo'] = df['principioactivo'].str.upper()
df['formafarmaceutica'] = df['formafarmaceutica'].str.upper()
df['descripcionatc'] = df['descripcionatc'].str.upper()
df['unidadmedida'] = df['unidadmedida'].str.upper()
df['estadoregistro'] = df['estadoregistro'].str.upper()
df.head()
```

```python
print("Manejo de valores nulos")
print("Conteo de columnas antes de limpiar valores nulos")
columnas_antes = len(df.columns)
print(f"Cantidad de columnas antes de la limpieza: {columnas_antes}")
porcentaje_nulos = (df.isnull().sum() / len(df)) * 100

print("Mostrar columnas con valores nulos:")
print("Porcentaje de valores nulos por columna:")
for columna, porcentaje in porcentaje_nulos.items():
    if porcentaje > 0:
        print(f"{columna}: {porcentaje:.2f}%")

print("Eliminar columnas con más de 50% de valores nulos")
df_limpio = df.loc[:, porcentaje_nulos <= 50]

print("Conteo de columnas despues de limpiar valores nulos")
columnas_despues = len(df_limpio.columns)
print(f"cantidad de columnas después de la limpieza: {columnas_despues}")
print(f"Columnas que se eliminaron: {columnas_antes - columnas_despues}")
print()

print("Columnas que quedaron:")
print(df_limpio.columns.tolist())
```

```python
print("Filtrado por top 10 de categorías terapéuticas más importantes")
top_10_categorias = df_limpio['descripcionatc'].value_counts().head(10)

print("Categorías incluidas:")
print(top_10_categorias)
print()

df_filtrado = df_limpio[df_limpio['descripcionatc'].isin(top_10_categorias.index)]
print(f"Registros en top 15 categorías: {len(df_filtrado)}")
```
