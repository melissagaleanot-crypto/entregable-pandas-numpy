```python
import pandas as pd

dataframe_url = 'https://www.datos.gov.co/resource/sfcd-b3ey.json'
df = pd.read_json(dataframe_url)
df.head()

```

```python
#explorar datos...

df.tail() # obtiene los datos de la cola del df
```

```python
# eliminando propiedades de EPM con valor = a 0.00

df = df[df['propiedadepm'] != 0]
df.tail() #Si comparamos el primer tail() las propiedadesepm no contienen valores equivalentes a cero
```

```python
# Obtener datos por columnas especificas (Filtrado)

df_colums_unique = df[['tipodeinformacion','rangodeconsumo', 'propiedaddelcliente']]
df_colums_unique_filter = df_colums_unique[df_colums_unique['rangodeconsumo'] != "N/A"]
df_colums_unique_filter.head()
```

```python
# Obtener de df_colums_unique_filter la suma de las propiedades de cliente superiores a 150 y menores a 160 con tipo de consumo punta y tipo de informacion horaria

df_colums_unique_filter.query('propiedaddelcliente > 150 and propiedaddelcliente < 160 and rangodeconsumo == "Punta" and tipodeinformacion == "Horaria"')
```

```python
#Organizar valores del DF inicial en orden decendente y posteriormente retirar los que no se encuentran en el rando de propiedaddelcliente determinado

df_sorted = df.sort_values(by=['propiedaddelcliente'], ascending=False)
df_sorted.query("propiedaddelcliente > 100 and propiedaddelcliente < 140")
```

```python
#df punta diferente de 0

df_shared = df[df['compartido'] > 150 ]
df_rango = df[df['rangodeconsumo'] == 'Punta']
df_rango.merge(df_shared, on='tipodeinformacion')
```
